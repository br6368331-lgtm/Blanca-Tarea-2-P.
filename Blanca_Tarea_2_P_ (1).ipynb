```python
import random

# Simulación de 10,000 lanzamientos
n_simulaciones = 10000
exitos = 0

for _ in range(n_simulaciones):
    dado1 = random.randint(1, 6)
    dado2 = random.randint(1, 6)
    if dado1 + dado2 == 5:
        exitos += 1

prob_estimada = exitos / n_simulaciones
print(f"Frecuencia estimada (Simulación): {prob_estimada:.4f}")
```

```python
import random


def generar_pin():

    return [random.randint(0, 9) for _ in range(4)]


n_simulaciones = 10000
sin_repetir = 0

for _ in range(n_simulaciones):
    pin = generar_pin()

    if len(set(pin)) == 4:
        sin_repetir += 1

prob_estimada = sin_repetir / n_simulaciones
print(f"Probabilidad estimada sin repetición: {prob_estimada:.4f}")
```

```python
def lanzar_dado():
    return random.randint(1, 6)


n_ensayos = 20000
conteo_9 = 0
conteo_10 = 0

for _ in range(n_ensayos):
    suma = lanzar_dado() + lanzar_dado() + lanzar_dado()
    if suma == 9:
        conteo_9 += 1
    elif suma == 10:
        conteo_10 += 1

print(f"Frecuencia de la suma 9:  {conteo_9 / n_ensayos:.4f}")
print(f"Frecuencia de la suma 10: {conteo_10 / n_ensayos:.4f}")
```

```python
import random

# Urna: 3 rojas ('R') y 2 azules ('A')
urna = ["R", "R", "R", "A", "A"]
n_extracciones = 2000
azules = 0

for _ in range(n_extracciones):
    bola = random.choice(urna)
    if bola == "A":
        azules += 1

prob_estimada = azules / n_extracciones
print(f"Probabilidad de obtener bola azul: {prob_estimada:.4f}")
```

```python
import random


def lanzar_moneda():
    return random.choice(["águila", "sol"])


n_series = 10000
series_favorables = 0

for _ in range(n_series):

    serie = [lanzar_moneda() for _ in range(20)]


    racha_maxima = 0
    racha_actual = 0
    for resultado in serie:
        if resultado == "águila":
            racha_actual += 1
            if racha_actual > racha_maxima:
                racha_maxima = racha_actual
        else:
            racha_actual = 0

    if racha_maxima >= 5:
        series_favorables += 1

prob_estimada = series_favorables / n_series
print(f"Probabilidad de tener al menos una racha de 5 águilas: {prob_estimada:.4f}")
```

```python
def simular_atencion_tienda(num_clientes=500):
    tiempo_actual = 0.0
    fin_atencion_anterior = 0.0
    espera_mas_10min = 0

    for _ in range(num_clientes):

        interllegada = random.uniform(1, 5)

        tiempo_servicio = random.uniform(2, 4)

        tiempo_llegada = tiempo_actual + interllegada
        tiempo_actual = tiempo_llegada


        inicio_atencion = max(tiempo_llegada, fin_atencion_anterior)
        tiempo_espera = inicio_atencion - tiempo_llegada

        if tiempo_espera > 10:
            espera_mas_10min += 1

        fin_atencion_anterior = inicio_atencion + tiempo_servicio

    return espera_mas_10min / num_clientes



probabilidad_espera = simular_atencion_tienda(500)
print(f"Probabilidad de que un cliente espere más de 10 minutos: {probabilidad_espera:.4f}")
```
